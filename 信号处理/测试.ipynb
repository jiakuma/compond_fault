```python
import numpy as np
import matplotlib.pyplot as plt

# 设置支持中文显示的字体（如果你的环境不需要可注释掉）
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 1. 定义基本参数
t = np.linspace(0, 2, 1000)  # 时间范围 0 到 2 个周期
f0 = 1                       # 基频设为 1 Hz
w0 = 2 * np.pi * f0          # 角频率

# 2. 准备画布
plt.figure(figsize=(12, 8))

# --- 第一部分：分别绘制前几个奇次谐波 ---
plt.subplot(2, 1, 1)
for n in [1, 3, 5, 7]:
    # 按照公式：振幅为 1/n，频率为 n*w0
    y_n = (1/n) * np.sin(n * w0 * t)
    plt.plot(t, y_n, label=f'频率 {n}ω0, 振幅 1/{n}')

plt.title('图分解：各个独立的奇次正弦波')
plt.xlabel('时间 t')
plt.ylabel('振幅')
plt.legend(loc='upper right')
plt.grid(True)

# --- 第二部分：观察叠加（合成）过程 ---
plt.subplot(2, 1, 2)

# 仅基波
y_sum = np.sin(w0 * t)
plt.plot(t, y_sum, label='仅基波 (1ω0)', linestyle='--')

# 叠加前 4 项 (1, 3, 5, 7)
for n in [3, 5, 7]:
    y_sum += (1/n) * np.sin(n * w0 * t)
plt.plot(t, y_sum, label='叠加至 7ω0 (图中描述的程度)')

# 叠加更多项以观察极限情况 (例如叠加到第 99 次谐波)
for n in range(9, 100, 2):
    y_sum += (1/n) * np.sin(n * w0 * t)
plt.plot(t, y_sum, label='叠加至 99ω0 (高度逼近方波)', color='red', linewidth=2)

plt.title('图合成：正弦波叠加投射到时域的结果')
plt.xlabel('时间 t')
plt.ylabel('合成振幅')
plt.legend(loc='upper right')
plt.grid(True)

plt.tight_layout()
plt.show()
```
